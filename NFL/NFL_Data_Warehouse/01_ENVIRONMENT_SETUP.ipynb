## Check All Important Packages

```python
packages = [
    "pandas",
    "polars",
    "numpy",
    "duckdb",
    "requests",
    "httpx",
    "bs4",
    "pyarrow",
    "nflreadpy",
    "sklearn",
    "torch",
    "plotly"
]

import importlib.util
import pandas as pd
import duckdb

package_status = []

for package in packages:

    installed = (
        importlib.util.find_spec(package)
        is not None
    )

    package_status.append({
        "package": package,
        "installed": installed
    })

package_df = pd.DataFrame(package_status)

display(package_df)
```
